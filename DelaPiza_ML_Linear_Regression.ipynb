```python
#imports
import pandas as pd
import numpy as np
```

```python
#read data set
data = pd.read_csv("/winequality-red_dataset.csv")
```

```python
#check data
data.head()
```

```python
#mount drive
from google.colab import drive
drive.mount('/content/drive')
```

```python
#define features
x = data.drop(['fixed acidity', 'alcohol', 'density'], axis=1).values
y = data['pH'].values
```

```python
#check data
x
```

```python
#split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.5)
```

```python
#plotting using seaborn(pairplot)
import seaborn as sns
%matplotlib inline

sns.pairplot(data, x_vars = ['fixed acidity', 'alcohol', 'density'], y_vars = ['pH'], aspect = 1, kind = 'reg')
```

```python
#using matplotlib
import matplotlib.pyplot as plt
data.plot.scatter( x='fixed acidity', y='pH'),
data.plot.scatter( x='alcohol', y='pH'),
data.plot.scatter( x='density', y='pH')

plt.grid()
```

```python
#build the model
from sklearn.linear_model import LinearRegression
model = LinearRegression()
model.fit(x_train, y_train)
```

```python
#predict predict
y_pred = model.predict(x_test)
print(y_pred)
```

```python
#calc slope (coefficient) -> B1model.coef_
model.coef_
```

```python
#intercept -> B0
model.intercept_
```

```python

#evaluate model
from sklearn.metrics import r2_score
r2_score(y_test, y_pred)
```

```python
plt.figure(figsize=(10, 5))
plt.scatter(y_test, y_pred, color='blue')
plt.xlabel = 'Actual'
plt.ylabel = 'Predicted'
plt.title = 'Actual vs Prediction'
```
